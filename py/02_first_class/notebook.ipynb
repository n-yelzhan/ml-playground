Describe a telecom subscriber as a Python class with attributes and calculation methods.

```python
class Subscriber:
    def __init__(self, name, tariff, monthly_fee):
        self.name = name
        self.tariff = tariff
        self.monthly_fee = monthly_fee
    def cost_for(self, months):
        return self.monthly_fee * months
```

```python
sub = Subscriber('Aliya', 'Smart', 3990)
print(sub.cost_for(6))
```

```python
class Subscriber:
    def __init__(self, name, tariff, monthly_fee):
        self.name = name
        self.tariff = tariff
        self.monthly_fee = monthly_fee
        self.payments =[]
    def add_payment(self, amount):
        self.payments.append(amount)
    def arpu(self):
        if not self.payments:
            return 0
        return sum(self.payments)/len(self.payments)
    def debt(self):
        expected = self.monthly_fee * len(self.payments)
        return expected - sum(self.payments)
```

```python
sub = Subscriber('Aliya', 'Smart', 3990)
sub.add_payment(3990)
sub.add_payment(3990)
sub.add_payment(3990)
sub.add_payment(2990)

print('Payments:', sub.payments)
print('ARPU:', round(sub.arpu(),2))
print('Debt:', sub.debt())
```

```python
sub2 = Subscriber("Berik", "Unlim", 5990)
sub2.add_payment(5990)

print(sub.name, sub.payments, sub.debt())
print(sub2.name, sub2.payments, sub2.debt())
```

```python
print(type(sub), type(sub2))
print(type(sub) is type(sub2))
print(sub is sub2)
```

```python
class BadSubscriber:
    payments = []

    def __init__(self, name):
        self.name = name

    def add_payment(self, amount):
        self.payments.append(amount)

a = BadSubscriber("Aliya")
b = BadSubscriber("Berik")
a.add_payment(3990)

print(a.payments)
print(b.payments)
```

# What I learned

A class is a template; objects (instances) are filled-in copies with their own data.
self points to the specific object, so methods calculate on that object’s data.
Mutable attributes (like lists) must be created in __init__ via self, otherwise all objects share one list.
