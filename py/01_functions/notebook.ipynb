# Task

Write reusable functions for telecom and bank metrics (ARPU, churn flag, loan payment) instead of copy-pasting formulas.

# Function

```python
def arpu(revenue, subscribers) -> float:
    """Average revenue per user. Returns None if there are no subscribers."""
    if subscribers == 0:
        return None
    return revenue / subscribers

arpu(1_000_000, 2_500)
```

```python
def churn_flag(days_inactive, threshold=30):
    return days_inactive >= threshold

print(churn_flag(45))
print(churn_flag(20))
print(churn_flag(20, threshold=14))
```

```python
def loan_payment(amount: float, annual_rate: float, months: int=12) -> float:
    """Monthly annuity payment.

    amount: loan amount
    annual_rate: annual rate, e.g. 0.2 for 20%
    months: loan term in months
    """
    r = annual_rate / 12
    return amount * r / (1 - (1+r)** -months)

round(loan_payment(1_000_000, 0.2),2)
print(help(loan_payment))
```

```python
def total_revenue(*months):
    return sum(months)

print(total_revenue(100, 200, 150))
print(total_revenue(100, 200, 150, 300, 250))

def revenue_by_segments(**segments):
    for name, value in segments.items():
        print(f"{name}:{value}")
    return sum(segments.values())

revenue_by_segments(b2c=500, b2b=300, iot=50)
```

# Tests

```python
assert arpu(1000, 10) == 100
assert churn_flag(30)
assert not churn_flag(29)
assert total_revenue(100, 200) == 300
assert round(loan_payment(1_000_000, 0.2), 1) == 92634.5
assert abs(loan_payment(1_000_000, 0.2) - 92634.5) < 0.01

print("All tests passed")
```

```python
arpu(1000,0)
assert arpu(1000,0) is None
```

## 4. Result

- All tests passed, including edge cases (exactly 30 days, zero subscribers → None)

## 5. What I learned
1. Write functions with parameters and default values (churn threshold=30 can be changed to 14 in one call).
2. Type hints and docstrings document a function; help() shows them. assert checks the result; money must be compared with a tolerance, not ==.
3. *args collects positional arguments into a tuple, **kwargs collects named arguments into a dict.
4. Edge cases matter: exactly 30 days is churn; zero subscribers should return None, not 0 — 0 looks like a real number.
